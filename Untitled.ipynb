```python
"""
Figure: RF vs HGB comparison across streams, policies (all10 / rank1),
        and metrics (Macro MAE, Macro Within±5d).

Layout: 2 rows (metrics) × 2 columns (policies)  →  4 panels
Each panel: grouped bars per stream, one bar-pair per model (RF, HGB).

Paste the whole file into a notebook cell and call make_figure(DATA_DIR, OUTPUT_DIR),
or run the last two lines directly.
"""

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ── paths ──────────────────────────────────────────────────────────────────
DATA_DIR   = Path('/fs1/home/atikul/Phenology')
OUTPUT_DIR = DATA_DIR / 'phase3_outputs' / 'figures'

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ── constants ───────────────────────────────────────────────────────────────
STREAMS_ORDERED = ['S1_only', 'S2_only', 'Met_only', 'S1+S2', 'All']

STREAM_LABELS = {
    'S1_only' : 'S1',
    'S2_only' : 'S2',
    'Met_only': 'Met',
    'S1+S2'  : 'S1+S2',
    'All'    : 'All',
}

POLICY_DISPLAY = {
    'all10': 'Shared label',
    'rank1': 'Closest field',
}

METRICS = {
    'mae'     : ('Macro MAE (days)',   'mae'),
    'within_5d': ('Macro Within ±5 d', 'within_5d'),
}

# Colour palette — colourblind-safe (Paul Tol "muted")
CLR_RF  = '#88CCEE'   # light blue  – RF
CLR_HGB = '#44AA99'   # teal        – HGB

EDGECOL  = 'black'
LW_EDGE  = 0.5

# ── helpers ─────────────────────────────────────────────────────────────────

def load_and_agg(csv_path: Path) -> pd.DataFrame:
    """Load a results CSV and return per-(stream, policy, metric) means."""
    df = pd.read_csv(csv_path)
    agg = (df.groupby(['stream', 'policy'], observed=True)
             .agg(mae=('mae', 'mean'), within_5d=('within_5d', 'mean'))
             .reset_index())
    return agg


def stream_series(agg: pd.DataFrame, policy: str, metric: str) -> pd.Series:
    sub = agg[agg['policy'] == policy].set_index('stream')
    return sub[metric].reindex(STREAMS_ORDERED)


# ── main figure ─────────────────────────────────────────────────────────────

def make_figure(data_dir: Path, output_dir: Path) -> Path:
    rf_csv  = data_dir / 'phase3_outputs' / 'rf'  / 'rf_results_rf.csv'
    hgb_csv = data_dir / 'phase3_outputs' / 'rf'  / 'rf_results_hgb.csv'

    rf_agg  = load_and_agg(rf_csv)
    hgb_agg = load_and_agg(hgb_csv)

    policies = ['all10', 'rank1']
    metric_keys = list(METRICS.keys())          # ['mae', 'within_5d']

    n_streams = len(STREAMS_ORDERED)
    x = np.arange(n_streams)
    w = 0.36                                    # bar width

    # ── Figure: 2 rows (metrics) × 2 cols (policies) ──────────────────────
    fig, axes = plt.subplots(
        nrows=2, ncols=2,
        figsize=(7.0, 4.6),                     # fits a Word A4 column nicely
        sharey='row',
        constrained_layout=True,
    )

    for row, mkey in enumerate(metric_keys):
        ylabel, col = METRICS[mkey]
        for c, pol in enumerate(policies):
            ax = axes[row, c]

            rf_vals  = stream_series(rf_agg,  pol, mkey)
            hgb_vals = stream_series(hgb_agg, pol, mkey)

            bars_rf  = ax.bar(x - w/2, rf_vals,  w,
                              color=CLR_RF,  edgecolor=EDGECOL, linewidth=LW_EDGE,
                              label='RF', zorder=3)
            bars_hgb = ax.bar(x + w/2, hgb_vals, w,
                              color=CLR_HGB, edgecolor=EDGECOL, linewidth=LW_EDGE,
                              label='HGB', zorder=3)

            ax.set_xticks(x)
            ax.set_xticklabels([STREAM_LABELS[s] for s in STREAMS_ORDERED],
                               fontsize=8)
            ax.tick_params(axis='y', labelsize=8)
            ax.yaxis.grid(True, linewidth=0.4, linestyle='--',
                          color='#cccccc', zorder=0)
            ax.set_axisbelow(True)

            # ── y-axis label: only left column ───────────────────────────
            if c == 0:
                ax.set_ylabel(ylabel, fontsize=8.5)

            # ── column header: only top row ───────────────────────────────
            if row == 0:
                ax.set_title(POLICY_DISPLAY[pol], fontsize=9, fontweight='bold',
                             pad=4)

            # ── y-limits: add 15 % headroom ───────────────────────────────
            ymax = max(rf_vals.max(), hgb_vals.max())
            if mkey == 'mae':
                ax.set_ylim(0, ymax * 1.20)
            else:
                # within_5d is a fraction → cap at 1.0
                ax.set_ylim(0, min(ymax * 1.20, 1.0))

            # ── value labels on bars ──────────────────────────────────────
            for bar in list(bars_rf) + list(bars_hgb):
                h = bar.get_height()
                if np.isnan(h):
                    continue
                fmt = f'{h:.1f}' if mkey == 'mae' else f'{h:.2f}'
                ax.text(bar.get_x() + bar.get_width() / 2, h + ymax * 0.01,
                        fmt, ha='center', va='bottom', fontsize=5.5, color='#333')

    # ── shared legend below figure ─────────────────────────────────────────
    legend_patches = [
        mpatches.Patch(facecolor=CLR_RF,  edgecolor=EDGECOL,
                       linewidth=LW_EDGE, label='RF'),
        mpatches.Patch(facecolor=CLR_HGB, edgecolor=EDGECOL,
                       linewidth=LW_EDGE, label='HGB'),
    ]
    fig.legend(handles=legend_patches,
               loc='lower center',
               ncol=2,
               frameon=False,
               fontsize=8.5,
               bbox_to_anchor=(0.5, -0.04))

    # ── save ───────────────────────────────────────────────────────────────
    out_pdf = output_dir / 'F_rf_vs_hgb.pdf'
    out_png = output_dir / 'F_rf_vs_hgb.png'
    fig.savefig(out_pdf, dpi=300, bbox_inches='tight')
    fig.savefig(out_png, dpi=300, bbox_inches='tight')
    plt.close(fig)
    print(f'Saved → {out_pdf}')
    print(f'Saved → {out_png}')
    return out_pdf


if __name__ == '__main__':
    make_figure(DATA_DIR, OUTPUT_DIR)
```
